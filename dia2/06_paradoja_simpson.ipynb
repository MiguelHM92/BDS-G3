# PARADOJA DE SIMPSONS

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
penguins_df = sns.load_dataset('penguins')
penguins_df
```

## ELIMINAMOS NULOS

```python
penguins_df.dropna(inplace=True)
penguins_df
```

## CORRELACIÓN DE VARIABLES

```python
sns.clustermap(
    data=penguins_df.corr(numeric_only=True),
    cmap=sns.diverging_palette(20, 230, as_cmap=True),
    center = 0,
    vmin = -1,
    vmax = 1,
    linewidths = 0.5,
    cbar_kws={'shrink':0.5},
    annot=True
)
```

# PARADOJA SIMPSON

## ANALIZAMOS LA RELACIÓN TOAL ENTRE EL BILL_LENGTH_MM Y EL BILL_DEPTH_MM

```python
%matplotlib inline
sns.set_style(style='whitegrid')
sns.set_context(context='notebook')
plt.rcParams['figure.figsize'] = (11, 9.4)

penguin_color = {
    'Adelie': '#ff6602ff',
    'Gentoo': '#0f7175ff',
    'Chinstrap': '#c65dc9ff'
}
```

```python
sns.scatterplot(
    data=penguins_df,
    x='bill_length_mm',
    y='bill_depth_mm'
)
```

## APLICAMOS RELACIÓN PARCIAL (AGRUPAR LOS RESULTADOS POR LA SPECIE)

```python
sns.scatterplot(
    data=penguins_df,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',
    palette=penguin_color
)
```

## GRAFICAMOS UNA REGRESIÓN LINEAL SIMPLE PARA ESTA AGRUPACIÓN

```python
sns.lmplot(
    data=penguins_df,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',
    height=10,
    palette=penguin_color
)
```

## APLICAMOS LA AGRUPACIÓN PARA EL FLIPPER_LENGTH_MM Y BODY_MASS_G

```python
sns.lmplot(
    data=penguins_df,
    x='flipper_length_mm',
    y='body_mass_g',
    hue='species',
    height=10,
    palette=penguin_color
)
```

## APLICAMOS LA REGRESIÓN LINEAL MULTIPLE AGRUPANDOLA POR SPECIES

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

modelo_total = smf.ols("body_mass_g ~ flipper_length_mm", data=penguins_df).fit()
print("Regresión lineal del dataset completo:")
print(modelo_total.summary())

sns.lmplot(x='flipper_length_mm', y='body_mass_g', data=penguins_df)
plt.title('Regresión total')
plt.show()
```

## REGRESIÓN LINEAL PARCIAL

```python
#modelo parcial por species
species_models = {}
for species,df in penguins_df.groupby('species'):
    model = smf.ols("body_mass_g ~ flipper_length_mm", data=df).fit()
    species_models[species] = model
    print(f"Regresión lineal para {species}:")
    print(model.summary())
    sns.lmplot(x='flipper_length_mm', y='body_mass_g', data=df)
    plt.title(f'Regresión para {species}')
    plt.show()
```
