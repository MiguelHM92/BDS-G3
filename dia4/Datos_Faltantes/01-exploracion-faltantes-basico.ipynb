```python
import pandas as pd
```

# CREAMOS UN DATAFRAME DE PRUEBA

```python
df = pd.DataFrame(
    data= {
        "a":list('asdfasdfas'),
        "b":range(0,10)
    }
)
df
```

# GENERAMOS VALORES NULOS

```python
df.iloc[2:5,0] = None
df.iloc[6:7,1] = None
df
```

# VERIFICAMOS VALORES NULO

```python
df.isna()
```

# CONTEO DE NULOS

```python
df.isna().sum()
```

# CONTEO TOTAL DE NULOS

```python
df.isna().sum().sum()
```

# CONTEO POR CASOS

```python
df.isna().pipe(
    lambda df_1: (
        df_1.sum()
        .reset_index(name='n_missing')
        .rename(columns={'index':'variable'})
        .assign(
            n_cases=len(df_1),
            pct_missing=lambda df_2: df_2.n_missing / df_2.n_cases * 100,
        )
    )
)
```
