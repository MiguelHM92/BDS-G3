```python
%matplotlib inline
sns.set_style(style='whitegrid')
sns.set_context(context='notebook')
plt.rcParams['figure.figsize'] = (11, 9.4)

penguin_color = {
    'Adelie': '#ff6602ff',
    'Gentoo': '#0f7175ff',
    'Chinstrap': '#c65dc9ff'
}
```

# EJERCICIO ANALISIS DE TOTAL_BILL Y TIP, Y TIP CON DAY USANDO TODO LO APRENDIDO

```python
tips_df = sns.load_dataset('tips')
tips_df
```

## 01. ANALISIS ENTRE TOTAL_BILL Y TIP (VARIABLES NUMÉRICAS):

```python
sns.scatterplot(
    data=tips_df,
    x='tip',
    y='total_bill',
    alpha=1/2,
    s=100
)
```

```python
sns.displot(
    data=tips_df,
    x='tip',
    y='total_bill',
    rug=True
)
```

```python
sns.displot(
    data=tips_df,
    x='tip',
    y='total_bill',
    rug=True,
    kind='kde'
)
```

## 02. ANÁLISIS DE TIP Y DAY (NUMERICA Y CATEGORICA):

```python
sns.stripplot(
    data=tips_df,
    x='day',
    y='tip',
    hue='day'
)
```

### Usando BOXPLOT:

```python
sns.boxplot(
    data=tips_df,
    x='tip',
    y='day',
    whis=np.inf
)
sns.stripplot(
    data=tips_df,
    x='tip',
    y='day',
    color='.3'
)
```

### Usando VIOLINPLOT:

```python
sns.violinplot(
    data=tips_df,
    x='day',
    y='tip',
    color='.8'
)
sns.stripplot(
    data=tips_df,
    x='day',
    y='tip',
    hue='day'
)
```

### Usando SWARMPLOT:

```python
sns.swarmplot(
    data=tips_df,
    x='day',
    y='tip',
    hue='day'
)
```
